```python


def story(path_):
     with open(path_) as f:
        return f.read()
path = '/Users/jordanjones/Documents/sherlock'
corpus=[]
import os
for filename in os.listdir(path):#String directory in list
    file_path = os.path.join(path, filename) #Path = path + filename
    text = story(file_path)
    corpus.append(text)



```

```python

import re
from nltk.stem import WordNetLemmatizer
import nltk
stemmer = WordNetLemmatizer()

en_stop = set(nltk.corpus.stopwords.words('english'))

def clean_text(document):
        document = document.lower() #lower case
        document = re.sub('\n+', ' ', document ) #Get rid of the string \n
        document = re.sub("'", "", document) #Prevent don't becoming don t etc
        document = re.sub('[^\w\s+]', ' ', document) #Remove anything that is not digits or letters and then remove whitespace
        document = re.sub(r'\s+', ' ', document) #Reomove all whitepaces
        words = document.split()
        words = [stemmer.lemmatize(word) for word in words]
        words = [word for word in words if word not in en_stop]
        words =[word for word in words if len(word) > 5]
      
        
        return words
 # add your code
formatted_data = []
for doc in corpus:
    words = clean_text(doc)
    formatted_data.append(words)
```

```python
from gensim import corpora
gensim_dict = corpora.Dictionary(formatted_data)
gensim_corpus = [gensim_dict.doc2bow(word) for word in formatted_data]

print(formatted_data[0])
print(gensim_corpus[0])
len(gensim_corpus)
```

```python
import gensim.models
from gensim.models import TfidfModel
model = TfidfModel(gensim_corpus)
```

```python

large_corpus=[]
for i in range(len(gensim_corpus)):
    story_words = []
    weights = model[gensim_corpus[i]]
    sorted_tfidf_weights = sorted(weights, key=lambda w: w[1], reverse=True) 
    #This paticular block is from a DataCamp course:
    #weights is a tuple of the token id and its tf-idf score. The lambda call makes sure sorting
    #is done on the right side of the tuple and reverse is another version of 'ascending = False.' 
    for term_id, weight in sorted_tfidf_weights[:200]:
        #Then to get top 50 scores, as they are already sorted, just call the previous dict made to get
        #the term id.

        word = gensim_dict.get(term_id)
        story_words.append(word)
    large_corpus.append(story_words)

len(large_corpus)




```

```python



from gensim import corpora
gensim_dict = corpora.Dictionary(large_corpus)
gensim_corpus = [gensim_dict.doc2bow(word) for word in large_corpus]

lda_topic_models=gensim.models.ldamodel.LdaModel(gensim_corpus, num_topics=68, id2word=gensim_dict, passes=20)
lda_topcis = lda_topic_models.print_topics(num_words=5)
for topic_name in lda_topcis:
    print(topic_name)
```
